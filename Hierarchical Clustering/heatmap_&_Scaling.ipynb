# **Heatmaps to Visualize correlation matrix**

*Why are heatmaps important?*

A correlation matrix helps identify relationships between variables, detect redundant features, and select useful features, leading to simpler and more effective machine learning models.

```python
df_1 = sns.load_dataset('penguins')
```

```python
df_1.head(5)
```

```python
df_1.corr(numeric_only=True).round(2)
```

```python
sns.heatmap(df_1.corr(numeric_only=True).round(2), annot=True, vmin=-1, vmax=1, center=0, cmap='vlag')
```

```python
import numpy as np

matrix = df_1.corr(numeric_only=True).round(2)
mask = np.triu(np.ones_like(matrix, dtype=bool))
mask
```

```python
sns.heatmap(df_1.corr(numeric_only=True).round(2), annot=True, vmin=-1, vmax=1, center=0, cmap='vlag', mask=mask)
```

# **Min Max Scaler**

```python
import pandas as pd

data = {
    'Age': [25, 45, 27, 48, 24, 46],
    'Annual_Income': [40000, 80000, 42000, 82000, 39000, 81000]
}
```

```python
df_2 = pd.DataFrame(data)
```

```python
df_2
```

```python
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import StandardScaler
```

```python
scaler = MinMaxScaler()
```

```python
scaler_again = StandardScaler()
```

```python
scaled_features_again = scaler_again.fit_transform(df_2)
```

```python
df_scaled_again = pd.DataFrame(scaled_features_again, columns=['Age_Scaled_SS', 'Income_Scaled_SS'])
```

```python
df_scaled_again
```

Use StandardScaler when:
Features follow a roughly normal distribution.

Using algorithms like:

1.   Logistic Regression
2.   Linear Regression
3.   SVM
4.   PCA
5.   K-Means (commonly preferred)


Use MinMaxScaler when:
You need all features in the same fixed range (0–1).

Using:

1.   Neural Networks
2.   KNN
3.   Image pixel values
4.   Algorithms requiring bounded inputs

```python
scaled_features = scaler.fit_transform(df_2)
```

```python
df_scaled = pd.DataFrame(scaled_features, columns=['Age_Scaled', 'Income_Scaled'])
```
