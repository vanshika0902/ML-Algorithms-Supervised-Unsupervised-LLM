# **Simple Text Generation using Decoder-only Model GPT 2**

Decoder-only model is suitable for generating text or predicting next word in a sequence.

```python
!pip install transformers==4.51.3
```

```python
from transformers import pipeline
```

```python
generator = pipeline('text-generation', model='gpt2')
```

```python
result = generator("One upon a time, in a faraway land", max_length=30, num_return_sequences=1)
```

```python
print(result)
```

```python
print(result[0]['generated_text'])
```

# **Text Classification using Encoder-only Model BERT**

Encoder-only model are designed to understand the input but they don't generate next text.

```python
from transformers import pipeline

classifier = pipeline('sentiment-analysis')
```

```python
result = classifier("I love ice cream!")
```

```python
print(result)
```

# **Translation with Encoder-Decoder Model T5**

```python
from transformers import pipeline

translator = pipeline('translation_en_to_fr', model='t5-base')
```

```python
result = translator("Transformers are amazing!", max_length=40)
```

```python
print(result)
```

```python
print(result[0]['translation_text'])
```
